  FUNCTION pseudocode(x, y):

      SET moves TO empty list
      
      IF x <= 2 THEN
          APPEND "D" TO moves
      ENDIF
      
      IF x >= 0 THEN
          APPEND "L" TO moves
      ENDIF
      
      IF x < 2 THEN
          APPEND "R" TO moves
      ENDIF
      
      IF x > 0 THEN
          APPEND "U" TO moves
      ENDIF
      
      RETURN moves
  ENDFUNCTION



**Máy Hút Bụi (Model)**

```python
import random

def pseudocode(x, y):
    moves = []
    if x <= 2: moves.append("D")
    if x >= 0: moves.append("L")
    if x < 2:  moves.append("R")
    if x > 0:  moves.append("U")
    return moves

def model_based_agent():
    matrix = [[random.randint(0, 1) for _ in range(4)] for _ in range(4)]
    # Bộ nhớ (Model) lưu trữ trạng thái thế giới mà nó đã thấy
    model = [[None for _ in range(4)] for _ in range(4)]
    pos = [0, 0]

    print("--- MODEL-BASED REFLEX AGENT ---")

    for i in range(15):
        x, y = pos
        # Cập nhật Model (bộ nhớ)
        model[x][y] = matrix[x][y]

        if matrix[x][y] == 1:
            print(f"Bước {i+1}: Ô ({x}, {y}) bẩn -> Hút rác và cập nhật.")
            matrix[x][y] = 0
            model[x][y] = 0

        # Chiến lược Model-based: Ưu tiên di chuyển đến ô mà Model biết là có rác
        moves = pseudocode(x, y)
        move = random.choice(moves) # Đơn giản hóa: di chuyển khám phá

        if move == "D" and pos[0] < 3: pos[0] += 1
        elif move == "U" and pos[0] > 0: pos[0] -= 1
        elif move == "L" and pos[1] > 0: pos[1] -= 1
        elif move == "R" and pos[1] < 3: pos[1] += 1

    print("Kết thúc:")
    for row in matrix: print(row)

if __name__ == "__main__":
    model_based_agent()
```

**Máy Hút Bụi (Simple)**

```python
import random

def pseudocode(x, y):
    moves = []
    if x <= 2: moves.append("D")
    if x >= 0: moves.append("L")
    if x < 2:  moves.append("R")
    if x > 0:  moves.append("U")
    return moves

def simple_reflex_agent():
    # Khởi tạo ma trận 4x4 ngẫu nhiên
    matrix = [[random.randint(0, 1) for _ in range(4)] for _ in range(4)]
    pos = [random.randint(0, 3), random.randint(0, 3)]

    print("--- SIMPLE REFLEX AGENT ---")
    print("Trạng thái ban đầu:")
    for row in matrix: print(row)

    steps = 10
    for i in range(steps):
        x, y = pos
        # Simple Reflex: Chỉ nhìn ô hiện tại
        if matrix[x][y] == 1:
            print(f"Bước {i+1}: Ô ({x}, {y}) CÓ RÁC -> Hút rác")
            matrix[x][y] = 0
        else:
            print(f"Bước {i+1}: Ô ({x}, {y}) SẠCH.")

        # Di chuyển dựa trên pseudocode
        moves = pseudocode(x, y)
        move = random.choice(moves)

        # Cập nhật vị trí
        if move == "D" and pos[0] < 3: pos[0] += 1
        elif move == "U" and pos[0] > 0: pos[0] -= 1
        elif move == "L" and pos[1] > 0: pos[1] -= 1
        elif move == "R" and pos[1] < 3: pos[1] += 1

    print("Trạng thái sau 10 bước:")
    for row in matrix: print(row)

if __name__ == "__main__":
    simple_reflex_agent()
```

**8-Puzzle (Model)**

```python
import random

def pseudocode(x, y):
    moves = []
    if x <= 2: moves.append("D") # Trượt ô số từ dưới lên ô trống (ô trống đi xuống)
    if x >= 0: moves.append("L") # Trượt ô số từ trái sang ô trống (ô trống đi trái)
    if x < 2:  moves.append("R") # Trượt ô số từ phải sang ô trống (ô trống đi phải)
    if x > 0:  moves.append("U") # Trượt ô số từ trên xuống ô trống (ô trống đi lên)
    return moves

def puzzle_simulation():
    grid = [[1]*4 for _ in range(4)]
    blank_pos = [random.randint(0, 3), random.randint(0, 3)]
    grid[blank_pos[0]][blank_pos[1]] = 0

    print("--- 8-PUZZLE (4x4) MODEL SIMULATION ---")
    print(f"Vị trí ô trống ban đầu: {blank_pos}")

    for i in range(5):
        x, y = blank_pos
        valid_moves = pseudocode(x, y)
        move = random.choice(valid_moves)

        # Thực hiện trượt
        old_x, old_y = x, y
        if move == "D" and x < 3: x += 1
        elif move == "U" and x > 0: x -= 1
        elif move == "L" and y > 0: y -= 1
        elif move == "R" and y < 3: y += 1

        # Tráo đổi giá trị trên lưới
        grid[old_x][old_y], grid[x][y] = grid[x][y], grid[old_x][old_y]
        blank_pos = [x, y]
        print(f"Bước {i+1}: Di chuyển '{move}' -> Ô trống hiện tại: {blank_pos}")

if __name__ == "__main__":
    puzzle_simulation()
```
